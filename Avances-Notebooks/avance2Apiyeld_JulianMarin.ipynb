```python
# Encasulamos las claves y las cerramos para evitar que queden expuestas. 
datos = open("claves.txt")
claves = [linea.strip() for linea in datos]
datos.close()
```

```python
# Se define la URL base de la API de YELP y se configuran las claves para que se puedan acceder a los datos de la API sin dejar visibles las claves.
api_url='https://api.yelp.com/v3/businesses/search'
clientid= claves[0]
apikey= claves[1]
headers={'Authorization':'Bearer %s'%apikey}
```

```python
# Elijo la misma ciudad que en dataset del Avance1.
ciudad = "Miami" 
```

```python
# De esta manera se consulta el API de Yelp para obtener información sobre restaurantes en la ciudad especificada
for  i in range(0, 240, 40):
    params={'term':'restaurants','location': ciudad,'limit': 40, "offset": i}
    response=requests.get(api_url,params=params,headers=headers)
    data=response.json()
    print(data)
```

```python
# El primer elemento del diccionario indica el total de restaurants existentes en la API
print('En total la base de datos registra %d restaurants'%data['total'])
```

```python
# Consultamos el contenido de la data Businesses la cual contiene registros de restaurantes donde podemos observar nombre, id categorias, etc. 
data["businesses"]
```

```python
# creamos una lista de listas para luego crear el dataframe con 240 restaurantes  y poder analzarlo.

lista_listas= []
for  i in range(0, 240, 40):
    params={'term':'restaurants','location': ciudad,'limit': 40, "offset": i}
    response=requests.get(api_url,params=params,headers=headers)
    data=response.json()

    businesses = data.get("businesses", [])

    for business in businesses:
        buffer = []
        buffer.append(business.get('id', None))
        buffer.append(business.get('name', None))
        buffer.append(business.get('review_count', None))
   

        categories= business.get('categories', [])
        if categories and isinstance(categories, list):
            categoria_principal = categories[0].get('title', None)
        else:
            categoria_principal = None
        buffer.append(categoria_principal)

    
        buffer.append(business.get('rating', None))
        buffer.append(business.get('price', None))
        location = business.get('location', {})
        buffer.append(location.get('city', None))
        buffer.append(location.get('display_address', None))

        lista_listas.append(buffer)
```

```python
# consultamos la información de lista creada para obtener los data de la API.

lista_listas
```

```python
# creamos las columnas que va a registrar el Dataframe de la Api.
columnas=["id","name", "review_count","categoria_principal","rating", "price", "city","location_address"]
```

```python
# Creamos el Dataframe de la API  y lo declaramos con la variable df.
df= pd.DataFrame(lista_listas, columns=columnas)
df
```

```python
# Consultamos los valores presentes en la columna price para realizar la imputacion en los valores escogemos la moda ya que es una variable categorica.
df["price"].value_counts()
```

```python
# se le imputa el valor de moda a la columna price para reemplazar los valores nulos que se encuentran en la misma, de esta manera se evita perder registros  y además reemplazamos los valores de price para normalizar y realizar una integracion conceptual con la base de datos de restaurantes.
df["price"]=df["price"].fillna(df["price"].mode()[0])

mapa_precio= {
    "$" : "Bajo",
    "$$": "Medio",
    "$$$": "Alto",
    "$$$$": "Muy Alto"
}


df["price"] = df["price"].map(mapa_precio)
```

```python
# con la función unique revisamos la columna categoria_principal para filtrar por los restaurantes que ofrecen servicio de bar o bebidas alcoholicas.
df["categoria_principal"].unique()
```

```python
# Creamos la variable bares para agrupar a los restaurantes que ofrecen servicio de bar con sus respetivas reseñas y calificación podemos observar que se registran 25 restaurantes que ofrecen servicio de bar lo anterior lo realizamos para realizar la integracion conceptual con la base de datos de clientes.
bares = (df[df["categoria_principal"].str.contains(r"\bbars?\b|pub|brewer|cocktailos dewine|lounge", case=False, na=False)]
         .groupby(["name", "categoria_principal"])
         .agg(rating_promedio=("rating", "mean"),
              total_reviews=("review_count", "sum"))
         .round(2)
         .sort_values("total_reviews", ascending=False)
         .reset_index())

bares
```

```python
# Realizamos el cálculo del coeficiente de correlacion de pearson  y observamos que no existe corelación entre las reseñas y la calificación esta muy cerca de 0 se considera nula pero tambien tiene signo negativo lo que nos deja ver que a mayor reseñas disminuye la calificación.
df["review_count"].corr(df["rating"])
```

```python
# Importamos la librería scipy con su paquete spearmanr para calcular el coeficiente de correlación de spearmanr el cual es mas robusto y no se ve afectao por valores muy extremos lo que nos deja ver una relación negativa en cuanta mayor cantidad de rreseñas menor es la calificación.
from scipy.stats import spearmanr

datos = df[["rating", "review_count"]]
rho, p = spearmanr(datos["rating"], datos["review_count"])
print(f"rho = {rho:.3f}, p = {p:.4f}")
```
